# Sorting Data with BigQuery Public Dataset - CDC Births Data Summary

In this notebook, we’ll sort data by using SQL queries with ORDER BY and WHERE clauses. This will enable us to organize and use data more efficiently as a data analyst.

## Dataset
We’ll query the <b>CDC Births Data Summary public dataset</b>. \
We will write queries to obtain some information about which counties in the United States have the most and least births in the years 2016-2018. 

## Load the dataset

```python
from google.cloud import bigquery

print(bigquery.__version__)
```

```python
client = bigquery.Client()

print(client.project)
```

```python
# This code only tests if the BigQuery connection is on.
query = """
SELECT
  name,
  SUM(number) AS total_births
FROM `bigquery-public-data.usa_names.usa_1910_current`
GROUP BY name
ORDER BY total_births DESC
LIMIT 10
"""

df = client.query(query).to_dataframe()

df.head()
```

```python
# Load the data
query = """
SELECT 
  * 
FROM 
  `bigquery-public-data.sdoh_cdc_wonder_natality.county_natality` 
LIMIT 
  1000
"""

data = client.query(query).to_dataframe()
data.head()
```

## Use the ORDER BY clause

Now, imagine we were asked by our manager to figure out which 10 counties had the lowest birth count for 2016-2018. We could accomplish this by modifying our query to use the ORDER BY clause.

```python
query = """
SELECT 
  *
FROM 
  `bigquery-public-data.sdoh_cdc_wonder_natality.county_natality` 
ORDER BY
  Births
LIMIT 
  10
"""

data = client.query(query).to_dataframe()
data
```

We noticed that Tompkins County, NY, had just 735 births in 2018—the lowest birth count of any county in the US between 2016-2018.

### Use DESC to reverse sorting order

Now, we can also modify the query to sort in the other direction, returning the top 10 counties with the highest yearly birth counts between 2016-2018. 

```python
query = """
SELECT 
  *
FROM 
  `bigquery-public-data.sdoh_cdc_wonder_natality.county_natality` 
ORDER BY
  Births
DESC
LIMIT 
  10
"""

data = client.query(query).to_dataframe()
data
```

From the 10 rows with the largest values in the Birth column, we see that Los Angeles County takes up the top three spots.

## Combine ORDER BY with WHERE clauses

Next, we continue to modify the query so that it returns the top 10 counties with the highest birth counts for 2018 only. To do this, we add a WHERE clause to the query that specifies only rows that have a Year value equal to 2018-01-01. 

```python
query = """
SELECT 
  *
FROM 
  `bigquery-public-data.sdoh_cdc_wonder_natality.county_natality` 
WHERE 
  Year = '2018-01-01'
ORDER BY
  Births
DESC
LIMIT 
  10
"""

data = client.query(query).to_dataframe()
data
```

From the results, we can see that the Los Angeles County, CA still took the 1st position with 110,271 births in 2018.
